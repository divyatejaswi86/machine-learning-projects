```python
# @title Importing Important Dependencies
import numpy as np
import pandas as pd
```

```python
# Importing LogisticRegression Model
from sklearn.linear_model import LogisticRegression
```

```python
# Importing accuracy score function
from sklearn.metrics import accuracy_score
```

```python
# Importing train test split function
from sklearn.model_selection import train_test_split
```

```python
# To convert all the string format into numeric
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Importing the Dataset
df = pd.read_csv('blood.csv')
df
```

```python
df.info()
```

```python
# Checking for null values
df.isnull().sum()
```

```python
# Basic Analysis
df['Class'].value_counts().reset_index()
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
sns.lineplot(x = df['Recency'],y = df['Monetary'], hue = df['Class'])
plt.title("Analysis For Monetary  Value According to the Age")
plt.xlabel("Recency")
plt.ylabel("Monetary")
plt.show()
```

```python
# Seperating Input Feature and Label
# X - Input Feature(Every Single Coulmn except the Outcome column)
# Y - Label(Outcome)
X = df.drop(columns = 'Class', axis = 1)
Y = df['Class']
```

```python
print(X)
```

```python
print(Y)
```

```python
# Splitting the data into training data and testing data
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2)
print("Shape of Original Data",X.shape)
print("Shape of Training Data(80%)",X_train.shape)
print("Shape of Testing Data(20%)",X_test.shape)
```

```python
# Model Training
model = LogisticRegression()
# Training the ML model using training data
model.fit(X_train, Y_train)
```

```python
# Checking accuracy score
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
print("Accuracy Score of training data", training_data_accuracy)
```

```python
# Checking accuracy score
X_test_prediction = model.predict(X_test)
testing_data_accuracy = accuracy_score(Y_test, X_test_prediction)
print("Accuracy Score of testing data", testing_data_accuracy)
```

```python
# Building a predictive system
input_data = (2,5,1250,11)
input_data_as_numpy_array = np.asarray(input_data)
# Reshape the numpy array as one data point
input_data_reshape = input_data_as_numpy_array.reshape(1,-1)
```

```python
# Prediction for one data point
prediction = model.predict(input_data_reshape)
if prediction == 1:
  print("bad")
else:
  print("good")
```
