```python
import pandas as pd
import numpy as np
```

```python
# Importing LinearRegression Model
from sklearn.linear_model import LinearRegression
```

```python
# Importing accuracy score function
from sklearn import metrics
```

```python
# Imprrting train test split function
from sklearn.model_selection import train_test_split
```

```python
# Importing df
df = pd.read_csv('car data.csv')
df
```

```python
# Info of the dataset
df.info()
```

```python
# Checking the null values
df.isnull().sum()
```

```python
# Checking the distribution of ctaegorical data
pd.unique(df['Seller_Type'])
```

```python
# Checking the distribution of ctaegorical data
pd.unique(df['Fuel_Type'])
```

```python
# Checking the distribution of ctaegorical data
pd.unique(df['Transmission'])
```

```python
# Encoding the Fuel_Type column
df.replace({'Fuel_Type' : {'Petrol' : 0, 'Diesel' : 1, 'CNG' : 2}}, inplace = True)
```

```python
# Encoding the Seller_Type column
df.replace({'Seller_Type' : {'Dealer' : 0, 'Individual' : 1}}, inplace = True)
```

```python
# Encoding the Transmission column
df.replace({'Transmission' : {'Manual' : 0, 'Automatic' : 1}}, inplace = True)
```

```python
# Splitting the Data into Input Feature(X) and Label(Y)
X = df.drop(['Car_Name', 'Selling_Price'], axis = 1)
Y = df['Selling_Price']
```

```python
print(X)
```

```python
print(Y)
```

```python
# Splitting the data into training and testing data
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.1)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
# Model Training
model = LinearRegression()
```

```python
# Training the ML model using training data
model.fit(X_train, Y_train)

# Prediction of training data
training_data_prediction = model.predict(X_train)
# To find the accuracy score do not use accuracy_score function
training_data_accuracy = metrics.r2_score(Y_train, training_data_prediction)
print("Accuracy score of Training data is", training_data_accuracy)
```

```python
# Prediction of testing data
testing_data_prediction = model.predict(X_test)
# To find the accuracy score do not use accuracy_score function
testing_data_accuracy = metrics.r2_score(Y_test, testing_data_prediction)
print("Accuracy score of Testing data is", testing_data_accuracy)
```

```python
# Building a Predictive System
input_data = (2004,12.35,135154,0,0,1,0)
input_data_as_numpy_array = np.asarray(input_data)
# Reshape the numpy array as we are predicting for one data point
input_data_reshape = input_data_as_numpy_array.reshape(1,-1)
```

```python
prediction = model.predict(input_data_reshape)
print("Car Price according to the prediction is", prediction)
```
